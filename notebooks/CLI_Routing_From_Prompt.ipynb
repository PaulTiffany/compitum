## Example: Batch CLI (reference)

Equivalent shell commands (not executed here) to route multiple prompts and extract fields with jq.

```python
commands = '''
#!/bin/bash
PROMPTS=("Pythagorean summary" "Binomial proof" "Privacy draft")
for p in "${PROMPTS[@]}"; do
  echo "=== Route: $p ==="
  compitum route --prompt "$p" \
    --constraints configs/constraints_us_default.yaml \
    --defaults configs/router_defaults.yaml --trace \
    | jq --arg prompt "$p" '{prompt: $prompt, model: .model, U: .utility, feasible: .constraints.feasible}'
done > batch_traces.jsonl
'''
print(commands)
```
