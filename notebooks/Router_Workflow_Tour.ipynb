# Router Workflow Tour

This notebook gives a high-level tour of Compitum's routing workflow.
It is intentionally lightweight so it executes in CI (nbmake).

Topics:
- What the router consumes (features, predictors, constraints)
- What the router produces (selection + certificate)
- Offline, deterministic testing and artifacts

See `WORKFLOWS.md` for how heavy jobs (RouterBench, mutation) are gated.

## Inputs and Outputs (minimal)

- Inputs: prompt (or embedding), context, configuration
- Outputs: chosen model name and a JSON certificate

Below we import and construct a minimal SwitchCertificate for illustration.

```python
from compitum.router import SwitchCertificate
demo = SwitchCertificate(
    model='demo',
    utility=0.0,
    utility_components={'distance': 0.0, 'evidence': 0.0},
    constraints={'feasible': True},
    boundary_analysis={'margin': 0.0},
    drift_status={'trust_radius': 1.0},
    pgd_signature='deadbeef',
    timestamp=0.0,
)
print(demo.to_json())
```
