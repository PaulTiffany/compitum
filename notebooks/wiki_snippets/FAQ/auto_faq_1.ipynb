```python
# Snippet: Parse Trace for Common Flags
import json

with open("trace.json") as f:
    cert = json.load(f)

# Real schema: drift info lives under cert["drift"], not a top-level drift_signal
drift_ema = cert["drift"]["drift_ema"]
if abs(drift_ema) > 0.1:
    print("High drift EMA -- consider a smaller trust_radius (r0) when constructing LyapunovController.")

if not cert["constraints"]["feasible"]:
    print("Constraint violation -- inspect A x vs. b in your constraints YAML.")

# Boundary check: is_boundary is a key returned by BoundaryAnalyzer.analyze(),
# not a standalone function -- it is already computed for you in the certificate.
if cert["boundary"]["is_boundary"]:
    print("Tie detected -- winner and runner_up were close; consider a different constraints file or seed.")

# Pro Tip: Automate in pytest: assert cert["constraints"]["feasible"], "Issue repro'd."
```
