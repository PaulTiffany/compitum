```python
# Snippet: Parse Trace for Common Flags
import json
from compitum.boundary import is_boundary

with open('trace.json') as f:
    cert = json.load(f)

# Check signals
drift = cert['drift_signal']
if abs(drift) > 0.1:
    print("High drift—tighten trust_radius in config.")
if not all(s >= 0 for s in cert['slacks']):
    print("Constraint violation—inspect A·x vs. b.")

# Boundary check
if is_boundary(cert['utilities']):
    print("Tie detected—reroute with higher tau for diversity.")

# Pro Tip: Automate in pytest: assert cert['feasible'], "Issue repro'd."
```
