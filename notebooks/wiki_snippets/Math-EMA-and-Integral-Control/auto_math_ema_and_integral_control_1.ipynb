```python
import numpy as np

class SRMFController:
    """
    Radius modulation using filtered control.
    
    Combines average for noise reduction and accumulator for offset correction, keeping radii bounded.
    """
    
    def __init__(self,
                 alpha: float = 0.3,
                 gamma: float = 0.1,
                 rho: float = 0.99,
                 beta: float = 0.5,
                 kappa: float = 0.2,
                 r_min: float = 0.1,
                 r_max: float = 2.0):
        self.alpha = alpha
        self.gamma = gamma
        self.rho = rho
        self.beta = beta
        self.kappa = kappa
        self.r_min, self.r_max = r_min, r_max
        
        self.d_bar = 0.0
        self.I = 0.0
        self.r = 1.0
    
    def update(self, d_t: float) -> float:
        """
        Process distance and compute new radius.
        
        Args:
            d_t: Distance at t.
        
        Returns:
            Updated radius.
        """
        self.d_bar = self.alpha * d_t + (1 - self.alpha) * self.d_bar
        e_t = d_t - self.d_bar
        self.I = self.rho * self.I + self.gamma * e_t
        self.r = np.clip(
            self.r * np.exp(-self.beta * self.d_bar - self.kappa * self.I),
            self.r_min,
            self.r_max
        )
        return self.r

# Example
ctrl = SRMFController(alpha=0.5, gamma=0.1, rho=0.99)
print("Signal | Smoothed | Integral | Radius")
print("-------|----------|----------|-------")
for d in [0.2, 0.3, 0.1, 0.25, 0.22, 0.28, 0.15]:
    r = ctrl.update(d)
    print(f" {d:.2f} | {ctrl.d_bar:.3f} | {ctrl.I:+.3f} | {r:.3f}")
```
