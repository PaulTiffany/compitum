```python
#!/usr/bin/env python3
import json
import glob
import matplotlib.pyplot as plt
traces = []
for f in sorted(glob.glob("batch_results/trace_*.json")):
    with open(f) as trace_file:
        traces.append(json.load(trace_file))
steps = list(range(1, len(traces) + 1))
drifts = [t['drift']['drift_ema'] for t in traces]
entropies = [t['boundary']['entropy'] for t in traces]
utilities = [t['utility'] for t in traces]
fig, axes = plt.subplots(3, 1, figsize=(10, 8))
axes[0].plot(steps, drifts, 'o-', color='green')
axes[0].axhline(y=0, color='red', linestyle='--', alpha=0.5)
axes[0].set_ylabel('Drift Signal ΔE')
axes[0].set_title('Free Energy Descent')
axes[0].grid(alpha=0.3)
axes[1].plot(steps, entropies, 's-', color='blue')
axes[1].axhline(y=1.0, color='orange', linestyle='--', alpha=0.5)
axes[1].set_ylabel('Entropy H')
axes[1].set_title('Diversity Measure')
axes[1].grid(alpha=0.3)
axes[2].plot(steps, utilities, '^-', color='purple')
axes[2].set_ylabel('Utility U')
axes[2].set_xlabel('Route Number')
axes[2].set_title('Performance Over Time')
axes[2].grid(alpha=0.3)
plt.tight_layout()
plt.savefig('batch_results/convergence_plot.png', dpi=150)
print("Plot saved to batch_results/convergence_plot.png")
```
