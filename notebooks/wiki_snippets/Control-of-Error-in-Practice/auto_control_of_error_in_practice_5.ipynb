```python
# In coherence.py - weighted reservoir sampling
def add(self, x: np.ndarray, w: float) -> None:
    self.tot_w += w
    if len(self.buf) < self.k:
        self.buf.append((x.copy(), w))
    else:
        j = self.rng.integers(0, int(self.tot_w))
        if j < self.k:
            self.buf[j] = (x.copy(), w)
```
