```python
import numpy as np  
from scipy.special import softmax  
utilities = np.array([2.0, 1.8, 1.5, 1.2, 0.8])  # Top-heavy  
k = len(utilities)  
for tau in [0.1, 0.5, 1.0]:  # Low τ: Focused; High: Balanced  
    probs = softmax(utilities / tau)  
    entropy = -np.sum(probs * np.log(probs + 1e-10))  
    print(f"τ={tau}: Probs={probs[:3]}..., H={entropy:.2f} (target ~log({k})={np.log(k):.2f})")  
# Output:  
# τ=0.1: Probs=[0.993,0.006,0.001]..., H=0.04 (target ~1.61)  
# τ=0.5: Probs=[0.634,0.277,0.070]..., H=0.79 (target ~1.61)  
# τ=1.0: Probs=[0.389,0.248,0.123]..., H=1.30 (target ~1.61)  
```
