```python
# Example: Accessibility Wrapper for Plots (from src/compitum/plotting.py)
import matplotlib.pyplot as plt
from typing import Any

def accessible_plot(fig: plt.Figure, alt_text: str, caption: str) -> None:
    """Ensure WCAG-compliant visuals."""
    # Color-independent: Cycle through markers/linestyles
    markers = ['o', 's', '^', 'D']
    linestyles = ['-', '--', '-.', ':']
    for i, line in enumerate(fig.get_lines()):
        line.set_marker(markers[i % len(markers)])
        line.set_linestyle(linestyles[i % len(linestyles)])
    
    # Alt/caption for export
    fig.suptitle(caption, fontsize=10)
    # Save with metadata (e.g., for docs site)
    fig.savefig('output.png', metadata={'alt': alt_text})
```
