```python
def log_euclidean_mean(matrices: list) -> np.ndarray:
    """
    Compute Fréchet mean of SPD matrices in log-Euclidean framework.
    
    Args:
        matrices: List of SPD matrices (each D x D).
    
    Returns:
        Mean SPD matrix.
    """
    log_matrices = [logm(M) for M in matrices]
    mean_log = np.mean(log_matrices, axis=0)
    return expm(mean_log)

# Example
matrices = [np.eye(3) * (i + 1) for i in range(5)]
mean_M = log_euclidean_mean(matrices)
print(f"Mean trace: {np.trace(mean_M):.3f}")
```
