```python
import numpy as np
import matplotlib.pyplot as plt

def visualize_distance_warping(L, delta=1e-3, grid_size=20):
    """
    Visualize how SPD metric warps distances in 2D.
    
    Args:
        L: Low-rank factor (2 x r).
        delta: Regularization parameter.
        grid_size: Grid resolution.
    """
    metric = SymbolicManifoldMetric(L, delta)
    
    # Create grid
    x = np.linspace(-2, 2, grid_size)
    y = np.linspace(-2, 2, grid_size)
    X_grid, Y_grid = np.meshgrid(x, y)
    
    # Origin point
    origin = np.array([0.0, 0.0])
    
    # Compute distances
    d_M = np.zeros((grid_size, grid_size))
    d_euclidean = np.zeros((grid_size, grid_size))
    
    for i in range(grid_size):
        for j in range(grid_size):
            point = np.array([X_grid[i, j], Y_grid[i, j]])
            d_M[i, j] = metric.distance(origin, point)
            d_euclidean[i, j] = np.linalg.norm(point)
    
    # Plot comparison
    fig, axes = plt.subplots(1, 3, figsize=(15, 5))
    
    # Euclidean distances
    im1 = axes[0].contourf(X_grid, Y_grid, d_euclidean, levels=20, cmap='viridis')
    axes[0].set_title('Euclidean Distance')
    axes[0].set_xlabel('x')
    axes[0].set_ylabel('y')
    plt.colorbar(im1, ax=axes[0])
    
    # SPD distances
    im2 = axes[1].contourf(X_grid, Y_grid, d_M, levels=20, cmap='plasma')
    axes[1].set_title('SPD Warped Distance')
    axes[1].set_xlabel('x')
    axes[1].set_ylabel('y')
    plt.colorbar(im2, ax=axes[1])
    
    # Ratio
    ratio = d_M / (d_euclidean + 1e-10)
    im3 = axes[2].contourf(X_grid, Y_grid, ratio, levels=20, cmap='coolwarm')
    axes[2].set_title('Warping Ratio (d_M / d_euclidean)')
    axes[2].set_xlabel('x')
    axes[2].set_ylabel('y')
    plt.colorbar(im3, ax=axes[2])
    
    plt.tight_layout()
    plt.show()

# Example with anisotropic stretching
L_anisotropic = np.array([[2.0, 0.1],
                          [0.1, 0.5]])
# visualize_distance_warping(L_anisotropic)
print("Visualization code ready (uncomment to display plots)")
```
