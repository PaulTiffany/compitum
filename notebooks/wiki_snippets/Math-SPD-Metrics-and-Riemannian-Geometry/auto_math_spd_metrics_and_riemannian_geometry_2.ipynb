```python
from scipy.linalg import logm, expm

def geodesic_path(M1: np.ndarray, M2: np.ndarray, t: float) -> np.ndarray:
    """
    Compute point along geodesic between M1 and M2 on SPD manifold.
    
    Args:
        M1: Start SPD matrix (D x D).
        M2: End SPD matrix (D x D).
        t: Parameter in [0, 1], where 0 gives M1 and 1 gives M2.
    
    Returns:
        SPD matrix M(t) on geodesic.
    """
    M1_inv = np.linalg.inv(M1)
    log_term = logm(M1_inv @ M2)
    return M1 @ expm(t * log_term)

# Example
D = 5
M1 = np.eye(D)
M2 = np.diag([1, 2, 3, 4, 5])

# Midpoint on geodesic
M_mid = geodesic_path(M1, M2, t=0.5)
print(f"Midpoint eigenvalues: {np.linalg.eigvalsh(M_mid)}")
```
