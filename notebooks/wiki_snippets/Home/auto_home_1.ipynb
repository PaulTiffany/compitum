```python
from compitum import Router, Observers, Constraints, EmbeddingPool

# 1) Pool and constraints
pool = EmbeddingPool.from_defaults()              # small set of text models
constraints = Constraints(
    latency_s_max=2.0,
    cost_usd_max=0.01,
    diversity_k_min=2,
)

# 2) Observers (mechanistic signals)
observers = Observers.default()                   # energy, coherence (KDE), boundary, entropy

# 3) Router
router = Router(pool=pool, observers=observers, constraints=constraints)

# 4) Route a prompt
prompt = "Summarize the key differences between mitosis and meiosis for grade 9."
decision = router.route(prompt)

# 5) Read the certificate
print(decision.certificate.to_json(indent=2))
```
